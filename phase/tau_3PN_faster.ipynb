# Fast version (truncated power series): `_faster` notebook

Same inputs as `tau_3PN.ipynb` (cells above are unchanged). The slow cell

```
tau_3PN = ((xe_expr/dx_dt_final.subs({SO^2:0},{SO^3:0}).subs({x:x_3PN_faster})).subs({e_t:et_e0_NNLO_3PN_faster})
           .taylor((Xi,0),(e0,0),11))*(2/3)
```

followed by keeping `Xi^0 ... Xi^6` (`tau_3PN_final`) is replaced by exact truncated power-series algebra
(same machinery as `et_e0_5_faster.ipynb`):

1. Everything lives in one Laurent polynomial ring over `QQ` in `e0, Xi` plus every other symbol as a
   coefficient variable (`chi^(1/18)`, `x0^(1/2)`, `m`, `eta`, spins, flags, `pi`, `euler_gamma`, logs).
2. After **every** multiplication we drop terms of total degree in `(e0, Xi)` above `ORDER = 11`,
   the same cut as `.taylor((Xi,0),(e0,0),11)`.
3. `e_t = e0 * u` with `u = et_e0/e0` (u starts at `chi^(-19/18)`), so
   `e0^a e_t^b -> e0^(a+b) u^b` (u^b = truncated binomial series), `log(e_t) -> log(e0) + log(u)`.
   `e_t` includes the `Xi^k` factors, exactly like your `et_e0_NNLO_3PN_faster`.
4. `X = x(e0)` = `xe_expr` with `e_t` substituted, as a series.
5. `dx_dt_final` is expanded in `e_t` to `e_t^ORDER` (one single-variable Maxima `taylor`, about 1 s). Then
   `x^p -> X^p` and `log(x) -> log(X)` (truncated series), `e_t -> e0*u`. This is your `.subs({x:x_3PN_faster})` (`x_3PN_faster` = `xe_expr`).
6. `1/dx_dt` is a truncated geometric series, `tau = 2/3 * X / dx_dt`, then only `Xi^0 ... Xi^6` are kept.

Maxima's two-variable `taylor` is **not** used: with exponents like `e0^(138/19)/e_t^(24/19)` it returns wrong
coefficients. The check cell at the end compares against a one-variable expansion instead
(`Xi -> s t, e0 -> t`, `taylor` in `t`), which matches exactly.

Output: `tau_files_faster/tau_3PN.txt`, same format as `tau_files/tau_3PN.txt` from `tau_3PN.ipynb`.

````python
# =====================================================================================
#  Truncated power-series machinery: fast, exact replacement for symbolic .taylor
# =====================================================================================
import operator
from sage.symbolic.operators import add_vararg, mul_vararg
ADD = (operator.add, add_vararg)
MUL = (operator.mul, mul_vararg)
SPECIAL = (e0, e_t, Xi)          # handled separately, never coefficient atoms
def is_log(op):
    return hasattr(op, "name") and op.name() == "log"

# ---- 1) find every symbol / constant appearing in the inputs -------------------------
def collect_atoms(exprs):
    """roots[s] = d  means symbol s only appears as powers s^(k/d)  (e.g. chi -> 18, x0 -> 2).
    consts      = non-symbol atoms like pi, euler_gamma, log(2), log(x0), log(chi)."""
    roots, consts = {}, {}
    def walk(ex):
        if ex.is_numeric():
            return
        op = ex.operator()
        if op is None:                                   # symbol or constant (pi, ...)
            if ex.is_symbol():
                if ex not in SPECIAL:
                    roots.setdefault(ex, 1)
            else:
                consts[str(ex)] = ex
            return
        if op in ADD or op in MUL:
            for a in ex.operands():
                walk(a)
            return
        if op is operator.pow:
            base, ee = ex.operands()
            ee = QQ(ee)
            if base.is_symbol() and base not in SPECIAL:
                roots[base] = lcm(roots.get(base, 1), ee.denominator())
            else:
                walk(base)
            return
        if is_log(op):
            arg = ex.operands()[0]
            if arg not in (e0, e_t):                     # log(e0), log(e_t) handled separately
                consts[str(ex)] = ex
            return
        raise ValueError("unsupported function in %s" % ex)
    for ex in exprs:
        walk(SR(ex).expand())
    return roots, consts

def _name(s):
    return "".join(ch if ch.isalnum() else "_" for ch in s) + "_r"

def build_ring(exprs):
    """Flat Laurent polynomial ring over QQ in  e0, Xi  and all coefficient atoms."""
    global L, e0_s, Xi_s, ROOTS, ATOM_GEN, GEN_SR, LOGE0
    ROOTS, consts = collect_atoms(exprs)
    for s, d in ROOTS.items():                         # log(s) is needed for log(s^(k/d))
        if d > 1:
            consts.setdefault(str(log(s)), log(s))
    names = ["e0_s", "Xi_s"] + [_name(str(s)) for s in ROOTS] + [_name(k) for k in consts] + ["log_e0_r"]
    L = LaurentPolynomialRing(QQ, names)
    g = L.gens()
    e0_s, Xi_s = g[0], g[1]
    ATOM_GEN, GEN_SR = {}, [e0, Xi]
    i = 2
    for s, d in ROOTS.items():                         # generator = s^(1/d)
        ATOM_GEN[s] = g[i]; GEN_SR.append(s**(QQ(1)/d)); i += 1
    for k, ex in consts.items():
        ATOM_GEN[k] = g[i]; GEN_SR.append(ex); i += 1
    LOGE0 = g[i]; GEN_SR.append(log(e0))
    print("series ring variables:", names)

# ---- 2) SR <-> ring ----------------------------------------------------------------
def to_P(ex):
    """Exact conversion of a symbolic expression into the ring L."""
    ex = SR(ex)
    if ex.is_numeric():
        return L(QQ(ex))
    op, args = ex.operator(), ex.operands()
    if op is None:
        if ex == e0: return e0_s
        if ex == Xi: return Xi_s
        if ex.is_symbol(): return ATOM_GEN[ex]**ROOTS[ex]
        return ATOM_GEN[str(ex)]
    if op in ADD: return sum(to_P(a) for a in args)
    if op in MUL: return prod(to_P(a) for a in args)
    if op is operator.pow:
        base, ee = args[0], QQ(args[1])
        if base.is_symbol() and base in ROOTS:
            k = ee * ROOTS[base]
            assert k in ZZ, ex
            return ATOM_GEN[base]**ZZ(k)
        assert ee in ZZ, ex
        if ee >= 0:
            return to_P(base)**ZZ(ee)
        b = to_P(base)
        assert len(b.monomials()) == 1, ex                 # only invert a single monomial
        return b**ZZ(ee)
    if is_log(op):
        if args[0] == e0: return LOGE0
        return ATOM_GEN[str(ex)]
    raise ValueError("cannot convert %s" % ex)

def to_SR(f):
    """Ring L -> symbolic expression (chi^(1/18) etc. restored)."""
    out = SR(0)
    for m, q in f.dict().items():
        out += q * prod(GEN_SR[i]**m[i] for i in range(len(m)) if m[i])
    return out

# ---- 3) truncation -----------------------------------------------------------------
def keep(m, N):
    """N integer: total degree in (e0, Xi) <= N  (like .taylor((Xi,0),(e0,0),N)).
    N = (Ne0, NXi): e0 power <= Ne0 AND Xi power <= NXi (separate limits)."""
    if isinstance(N, tuple):
        return m[0] <= N[0] and m[1] <= N[1]
    return m[0] + m[1] <= N

def trunc(f, N):
    return L({m: q for m, q in f.dict().items() if keep(m, N)})

def xi_part(f, k):
    """The Xi^k part of f (Xi^k factor included), like f.coefficient(Xi,k)*Xi^k."""
    return L({m: q for m, q in f.dict().items() if m[1] == k})

def bigrade(f):
    """split f by its (e0 power, Xi power):  {(a, k): part of f with e0^a Xi^k}."""
    parts = {}
    for m, q in f.dict().items():
        parts.setdefault((m[0], m[1]), {})[m] = q
    return {g: L(d) for g, d in parts.items()}

def mul_trunc(f, g, N):
    """trunc(f*g, N), but only multiplies the pieces whose product survives the cut.
    N = None: no truncation (plain product)."""
    if N is None:
        return f * g
    F, G = bigrade(f), bigrade(g)
    out = L(0)
    for (a1, k1), p in F.items():
        for (a2, k2), r in G.items():
            if keep((a1 + a2, k1 + k2), N):
                out += p * r
    return out

# ---- 4) series of f^b and log(f) ---------------------------------------------------
def split_unit(f):
    """f = U * (1 + delta): U = (e0, Xi)-free part, must be a single monomial c*monomial (c rational)."""
    U = L({m: q for m, q in f.dict().items() if m[0] == 0 and m[1] == 0})
    assert len(U.monomials()) == 1, U
    return U, mul_trunc(f, U**-1, None) - 1

def _order_len(N):
    return sum(N) if isinstance(N, tuple) else N

def series_pow(f, b, N):
    """f^b truncated at order N:  U^b * sum_n binom(b, n) delta^n."""
    b = QQ(b)
    U, delta = split_unit(f)
    e = U.exponents()[0]
    assert all((x * b) in ZZ for x in e), (U, b)
    c = QQ(U.coefficients()[0])
    cb = c**b
    assert cb in QQ, (c, b)
    Ub = L({tuple(ZZ(x * b) for x in e): QQ(cb)})
    res, term = L(1), L(1)
    for n in range(1, _order_len(N) + 1):
        term = mul_trunc(term, delta, N) * (b - n + 1) / n
        if term == 0:
            break
        res += term
    return mul_trunc(Ub, res, N)

def log_of_unit(U):
    """log of a monomial in the root atoms, e.g. log(chi^(-19/18)) = -19/18 * log(chi)."""
    e = U.exponents()[0]
    out = L(0)
    for s, d in ROOTS.items():
        k = e[list(L.gens()).index(ATOM_GEN[s])]
        if k:
            out += QQ(k) / d * ATOM_GEN[str(log(s))]
    return out

def series_log(f, N):
    """log(f) truncated at order N:  log(U) + sum_n (-1)^(n+1) delta^n / n."""
    U, delta = split_unit(f)
    res, term = log_of_unit(U), L(1)
    for n in range(1, _order_len(N) + 1):
        term = mul_trunc(term, delta, N)
        if term == 0:
            break
        res += (-1)**(n + 1) * term / n
    return res

# ---- 5) the replacement for  A.taylor((e_t,0),(e0,0),tord).subs({e_t: et_sub}) ---------
def monomial_split(t):
    """t = coef * e0^a * e_t^b * log(e0)^p * log(e_t)^q  ->  (coef, a, b, p, q)."""
    a = b = QQ(0); p = q = 0
    coef = SR(1)
    facs = t.operands() if t.operator() in MUL else [t]
    for fct in facs:
        base, ee = (fct.operands()[0], fct.operands()[1]) if fct.operator() is operator.pow else (fct, 1)
        if base == e0:            a += QQ(ee)
        elif base == e_t:         b += QQ(ee)
        elif base == log(e0):     p += ZZ(ee)
        elif base == log(e_t):    q += ZZ(ee)
        else:                     coef *= fct
    return coef, a, b, p, q

def xe_series(A, et_sub, tord, N):
    """Series version of  A.taylor((e_t,0),(e0,0),tord).subs({e_t: et_sub})  at order N.
    Each term coef*e0^a*e_t^b (a+b integer) becomes coef*e0^(a+b)*u^b with u = e_t/e0,
    log(e_t) becomes log(e0) + log(u)."""
    assert all(m[0] >= 1 for m in et_sub.exponents()), "e_t must be divisible by e0"
    u = et_sub * e0_s**-1
    cache, out = {}, L(0)
    logu = None
    terms = SR(A).expand()
    terms = terms.operands() if terms.operator() in ADD else [terms]
    for t in terms:
        coef, a, b, p, q = monomial_split(t)
        if a + b > tord:                       # same cut as .taylor((e_t,0),(e0,0),tord)
            continue
        assert (a + b) in ZZ and a + b >= 0, (t, a, b)
        if b not in cache:
            cache[b] = series_pow(u, b, N)
        piece = mul_trunc(to_P(coef) * e0_s**ZZ(a + b), cache[b], N)
        if p:
            piece *= LOGE0**p
        if q:
            if logu is None:
                logu = series_log(u, N)
            for _ in range(q):
                piece = mul_trunc(piece, LOGE0 + logu, N)
        out += trunc(piece, N)
    return out

# ---- 6) dx/dt with x -> x(e_t) ------------------------------------------------------
def split_x(t):
    """t = coef * x^p * log(x)^q * e_t^b  ->  (coef, p, q, b)."""
    p = b = QQ(0); q = 0
    coef = SR(1)
    facs = t.operands() if t.operator() in MUL else [t]
    for fct in facs:
        base, ee = (fct.operands()[0], fct.operands()[1]) if fct.operator() is operator.pow else (fct, 1)
        if base == x:             p += QQ(ee)
        elif base == log(x):      q += ZZ(ee)
        elif base == e_t:         b += QQ(ee)
        else:                     coef *= fct
    return coef, p, q, b

def dxdt_series(D_et, X, et_sub, N):
    """Series of  D_et.subs({x: x(e_t)}).subs({e_t: et_sub}),  where X = x(e_t(e0)) as a series.
    D_et: polynomial in e_t (already expanded); x^p -> X^p and log(x) -> log(X) as truncated series."""
    u = et_sub * e0_s**-1
    ucache, xcache, out = {}, {}, L(0)
    logX = None
    terms = SR(D_et).expand()
    terms = terms.operands() if terms.operator() in ADD else [terms]
    for t in terms:
        coef, p, q, b = split_x(t)
        assert b in ZZ and b >= 0, t
        if b > N:
            continue
        if b not in ucache:
            ucache[b] = series_pow(u, b, N)
        if p not in xcache:
            xcache[p] = series_pow(X, p, N)
        piece = mul_trunc(to_P(coef) * e0_s**ZZ(b), ucache[b], N)
        piece = mul_trunc(piece, xcache[p], N)
        if q:
            if logX is None:
                logX = series_log(X, N)
            for _ in range(q):
                piece = mul_trunc(piece, logX, N)
        out += piece
    return out
````

````python
# ---- FAST: setup ----
import os, time
os.makedirs("tau_files_faster", exist_ok=True)
ORDER = 11                                   # total degree in (Xi, e0), as in .taylor((Xi,0),(e0,0),11)

t_f = time.time()
# dx/dt, with the same SO^2, SO^3 -> 0 as the original cell, expanded in e_t up to e_t^ORDER.
# Enough: every e_t comes with at least one e0 (e_t = e0*u), and nothing else has negative e0 powers.
D_sym = dx_dt_final.subs({SO^2:0},{SO^3:0}).expand()
assert D_sym.coefficient(Xi, 1) == 0
D_et = sum(D_sym.coefficient(Xi, k).taylor(e_t, 0, ORDER) * Xi^k for k in range(7))
D_et = D_et.log_expand('all').expand()      # log(4*x^(3/2)/x0) -> 2*log(2) + 3/2*log(x) - log(x0), etc.
assert D_et.degree(Xi) <= 6

# e_t(e0, Xi) with the Xi^k factors (= your et_e0_NNLO_3PN_faster)
et_sub = sum(et_e0_NNLO_3PN.coefficient(Xi, k) * Xi^k for k in range(7)).expand()
assert et_e0_NNLO_3PN.coefficient(Xi, 1) == 0 and xe_expr.coefficient(Xi, 1) == 0

build_ring([xe_expr, D_et, et_sub, chi])
et_P = to_P(et_sub)
print("setup time: %.1f s" % (time.time() - t_f), flush=True)
````
