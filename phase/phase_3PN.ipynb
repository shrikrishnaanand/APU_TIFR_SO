```python
import re
filename = "tau_files_faster/tau_3PN.txt"

with open(filename, "r") as f:
    content = f.read()


match = re.search(r"tau_3PN\s*=\s*(.*)", content, re.DOTALL)
if match:
    expr_str = match.group(1).strip()
    tau_expr = (((sage_eval(expr_str, locals=globals()))).expand()).subs({SO^2:0},{SO^3:0})
```

```python
# k/(1+k) with e_t(e0,x,x0) substituted, Xi^2..Xi^6, to e0^4 (made in K_files.ipynb)
filename = "K_files/k_term.txt"
with open(filename, "r") as f:
    content = f.read()
match = re.search(r"k_term\s*=\s*(.*)", content, re.DOTALL)
if match:
    k_term = sage_eval(match.group(1).strip(), locals=globals())
W=SR.wild(0)
k_term = k_term.expand().subs({SO^2:0},{SO^3:0}).subs({chi^W:x^(3/2*W)*x0^(-3/2*W)}).subs({chi:x^(3/2)*x0^(-3/2)}).expand()
k_1PN_expr  = k_term.coefficient(Xi,2)
k_15PN_expr = k_term.coefficient(Xi,3)
k_2PN_expr  = k_term.coefficient(Xi,4)
k_25PN_expr = k_term.coefficient(Xi,5)
k_3PN_expr  = k_term.coefficient(Xi,6)
```

```python
omega_factor=(1/(G*m*Xi^3))
```

```python
# chi = F/F0 = (x/x0)^(3/2)  =>  chi^p = x^(3p/2) / x0^(3p/2)   (p < 0 here, so x0 goes upstairs)
# .expand() first so every term is a single power chi^p (otherwise taylor groups terms, e.g. (a*chi^(19/9) + b)/chi^(62/9))
tau_N_expr=(tau_expr.coefficient(Xi,0).taylor(e0,0,4)).expand().subs({chi^(-8/3):x0^4/x^4},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-62/9):x0^(31/3)/x^(31/3)})
tau_1PN_expr=(tau_expr.coefficient(Xi,2).taylor(e0,0,4)).expand().subs({chi^(-2):x0^3/x^3},{chi^(-37/9):x0^(37/6)/x^(37/6)},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-56/9):x0^(28/3)/x^(28/3)},{chi^(-62/9):x0^(31/3)/x^(31/3)})
tau_15PN_expr=(tau_expr.coefficient(Xi,3).taylor(e0,0,4)).expand().subs({chi^(-5/3):x0^(5/2)/x^(5/2)},{chi^(-34/9):x0^(17/3)/x^(17/3)},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-53/9):x0^(53/6)/x^(53/6)},{chi^(-62/9):x0^(31/3)/x^(31/3)})
tau_2PN_expr=(tau_expr.coefficient(Xi,4).taylor(e0,0,4)).expand().subs({chi^(-4/3):x0^2/x^2},{chi^(-31/9):x0^(31/6)/x^(31/6)},{chi^(-37/9):x0^(37/6)/x^(37/6)},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-50/9):x0^(25/3)/x^(25/3)},{chi^(-56/9):x0^(28/3)/x^(28/3)},{chi^(-62/9):x0^(31/3)/x^(31/3)})
tau_25PN_expr=(tau_expr.coefficient(Xi,5).taylor(e0,0,4)).expand().subs({chi^(-1):x0^(3/2)/x^(3/2)},{chi^(-28/9):x0^(14/3)/x^(14/3)},{chi^(-34/9):x0^(17/3)/x^(17/3)},{chi^(-37/9):x0^(37/6)/x^(37/6)},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-47/9):x0^(47/6)/x^(47/6)},{chi^(-53/9):x0^(53/6)/x^(53/6)},{chi^(-56/9):x0^(28/3)/x^(28/3)},{chi^(-62/9):x0^(31/3)/x^(31/3)})
tau_3PN_expr=(tau_expr.coefficient(Xi,6).taylor(e0,0,4)).expand().subs({chi^(-2/3):x0/x},{chi^(-25/9):x0^(25/6)/x^(25/6)},{chi^(-31/9):x0^(31/6)/x^(31/6)},{chi^(-34/9):x0^(17/3)/x^(17/3)},{chi^(-37/9):x0^(37/6)/x^(37/6)},{chi^(-43/9):x0^(43/6)/x^(43/6)},{chi^(-44/9):x0^(22/3)/x^(22/3)},{chi^(-50/9):x0^(25/3)/x^(25/3)},{chi^(-53/9):x0^(53/6)/x^(53/6)},{chi^(-56/9):x0^(28/3)/x^(28/3)},{chi^(-62/9):x0^(31/3)/x^(31/3)},{log(chi):3/2*log(x)-3/2*log(x0)})
```

```python
phi_N = ((omega_factor*(3/2)*sqrt(x)*tau_N_expr).integrate(x)).expand()
t_N= (((3/2)*(1/x)*tau_N_expr).integrate(x)).expand()
```

```python
psi_N=(j*phi_N-omega_factor*x^(3/2)*j*t_N).expand()
```

```python
psi_N_2=((psi_N/psi_N.coefficient(e0,0)).expand())
```

```python
psi_N_3=psi_N_2.subs({x0^(19/3):x^(19/3)/chi^(38/9)}).subs({x0^(19/6):x^(19/6)/chi^(38/18)})
```

```python
show(psi_N_3)
```

```python
# 1PN from tau_1PN, same as Newtonian
phi_1PN = ((omega_factor*(3/2)*sqrt(x)*tau_1PN_expr).integrate(x)).expand()
t_1PN   = (((3/2)*(1/x)*tau_1PN_expr).integrate(x)).expand()
psi_1PN_tau = (j*phi_1PN-omega_factor*x^(3/2)*j*t_1PN).expand()

# periastron-advance correction: k(x) inside the phi integral, k(F) at the end point multiplies t_N (from f)
phi_k  = ((omega_factor*(3/2)*sqrt(x)*(k_1PN_expr*tau_N_expr).expand()).integrate(x)).expand()
psi_1PN_k = (-(j+n)*(phi_k-omega_factor*x^(3/2)*k_1PN_expr*t_N)).expand()

psi_1PN = psi_1PN_tau + psi_1PN_k
```

```python
# normalise by the same Newtonian leading term, then x0 -> x*chi^(-2/3); dividing by x gives P_2 of Eq. (C6)
W=SR.wild(0)
psi_1PN_2=((psi_1PN/psi_N.coefficient(e0,0)).expand().taylor(e0,0,4)).expand()
psi_1PN_3=(psi_1PN_2.subs({x0^W:x^W*chi^(-2/3*W)})).expand()
P2=(psi_1PN_3/x).expand()
show(P2.collect(e0))
```

```python
# 1.5PN from tau_15PN, same as Newtonian
phi_15PN = ((omega_factor*(3/2)*sqrt(x)*tau_15PN_expr).integrate(x)).expand()
t_15PN   = (((3/2)*(1/x)*tau_15PN_expr).integrate(x)).expand()
psi_15PN_tau = (j*phi_15PN-omega_factor*x^(3/2)*j*t_15PN).expand()

# periastron-advance correction: SO part of k at 1.5PN acting on tau_N
phi_k15 = ((omega_factor*(3/2)*sqrt(x)*(k_15PN_expr*tau_N_expr).expand()).integrate(x)).expand()
psi_15PN_k = (-(j+n)*(phi_k15-omega_factor*x^(3/2)*k_15PN_expr*t_N)).expand()

psi_15PN = psi_15PN_tau + psi_15PN_k
```

```python
# normalise, x0 -> x*chi^(-2/3); dividing by x^(3/2) gives P_3
psi_15PN_2=((psi_15PN/psi_N.coefficient(e0,0)).expand().taylor(e0,0,4)).expand()
psi_15PN_3=(psi_15PN_2.subs({x0^W:x^W*chi^(-2/3*W)}).subs({x0:x*chi^(-2/3)})).expand()
P3=(psi_15PN_3/x^(3/2)).expand()
show(P3.collect(e0))
```

```python
# 2PN from tau_2PN, same as Newtonian
phi_2PN = ((omega_factor*(3/2)*sqrt(x)*tau_2PN_expr).integrate(x)).expand()
t_2PN   = (((3/2)*(1/x)*tau_2PN_expr).integrate(x)).expand()
psi_2PN_tau = (j*phi_2PN-omega_factor*x^(3/2)*j*t_2PN).expand()

# periastron-advance correction: Xi^4 parts of k*tau (inside the integral) and k(F)*t (end point)
ktau_2PN = ((k_2PN_expr*tau_N_expr+k_1PN_expr*tau_1PN_expr).expand().taylor(e0,0,4)).expand()
phi_k2   = ((omega_factor*(3/2)*sqrt(x)*ktau_2PN).integrate(x)).expand()
psi_2PN_k = (-(j+n)*(phi_k2-omega_factor*x^(3/2)*(k_2PN_expr*t_N+k_1PN_expr*t_1PN))).expand()

psi_2PN = psi_2PN_tau + psi_2PN_k
```

```python
# normalise, x0 -> x*chi^(-2/3); dividing by x^2 gives P_4
psi_2PN_2=((psi_2PN/psi_N.coefficient(e0,0)).expand().taylor(e0,0,4)).expand()
psi_2PN_3=(psi_2PN_2.subs({x0^W:x^W*chi^(-2/3*W)}).subs({x0:x*chi^(-2/3)})).expand()
P4=(psi_2PN_3/x^2).expand()
show(P4.collect(e0))
```

```python
# 2.5PN from tau_25PN, same as Newtonian
phi_25PN = ((omega_factor*(3/2)*sqrt(x)*tau_25PN_expr).integrate(x)).expand()
t_25PN   = (((3/2)*(1/x)*tau_25PN_expr).integrate(x)).expand()
psi_25PN_tau = (j*phi_25PN-omega_factor*x^(3/2)*j*t_25PN).expand()

# periastron-advance correction: Xi^5 parts of k*tau and k(F)*t
ktau_25PN = ((k_25PN_expr*tau_N_expr+k_15PN_expr*tau_1PN_expr+k_1PN_expr*tau_15PN_expr).expand().taylor(e0,0,4)).expand()
phi_k25   = ((omega_factor*(3/2)*sqrt(x)*ktau_25PN).integrate(x)).expand()
psi_25PN_k = (-(j+n)*(phi_k25-omega_factor*x^(3/2)*(k_25PN_expr*t_N+k_15PN_expr*t_1PN+k_1PN_expr*t_15PN))).expand()

psi_25PN = psi_25PN_tau + psi_25PN_k
```

```python
# normalise, x0 -> x*chi^(-2/3); dividing by x^(5/2) gives P_5
psi_25PN_2=((psi_25PN/psi_N.coefficient(e0,0)).expand().taylor(e0,0,4)).expand()
psi_25PN_3=(psi_25PN_2.subs({x0^W:x^W*chi^(-2/3*W)}).subs({x0:x*chi^(-2/3)})).expand()
P5=(psi_25PN_3/x^(5/2)).expand()
show(P5.collect(e0))
```

```python
# 3PN from tau_3PN, same as Newtonian
phi_3PN = ((omega_factor*(3/2)*sqrt(x)*tau_3PN_expr).integrate(x)).expand()
t_3PN   = (((3/2)*(1/x)*tau_3PN_expr).integrate(x)).expand()
psi_3PN_tau = (j*phi_3PN-omega_factor*x^(3/2)*j*t_3PN).expand()

# periastron-advance correction: Xi^6 parts of k*tau and k(F)*t (tau, t have no Xi^1 part)
ktau_3PN = ((k_3PN_expr*tau_N_expr+k_2PN_expr*tau_1PN_expr+k_15PN_expr*tau_15PN_expr+k_1PN_expr*tau_2PN_expr).expand().taylor(e0,0,4)).expand()
phi_k3   = ((omega_factor*(3/2)*sqrt(x)*ktau_3PN).integrate(x)).expand()
psi_3PN_k = (-(j+n)*(phi_k3-omega_factor*x^(3/2)*(k_3PN_expr*t_N+k_2PN_expr*t_1PN+k_15PN_expr*t_15PN+k_1PN_expr*t_2PN))).expand()

psi_3PN = psi_3PN_tau + psi_3PN_k
```

```python
# normalise, x0 -> x*chi^(-2/3) (log(x0) -> log(x) - 2/3 log(chi)); dividing by x^3 gives P_6
psi_3PN_2=((psi_3PN/psi_N.coefficient(e0,0)).expand().taylor(e0,0,4)).expand()
psi_3PN_3=(psi_3PN_2.subs({log(x0):log(x)-2/3*log(chi)}).subs({x0^W:x^W*chi^(-2/3*W)}).subs({x0:x*chi^(-2/3)})).expand()
P6=(psi_3PN_3/x^3).expand()
show(P6.collect(e0))
```

```python
# save each PN piece: Psi_j^n = ... - 3j/(256 eta x^(5/2)) * sum_m P_m x^(m/2)   (P_1 = 0)
import os
os.makedirs("psi", exist_ok=True)
for name, P in [("P0",psi_N_3),("P2",P2),("P3",P3),("P4",P4),("P5",P5),("P6",P6)]:
    with open("psi/%s.txt" % name, "w") as f:
        f.write("# Expansion of %s: coefficient of x^(m/2) in Psi_j^n, to e0^4, in x, chi, e0, j, n\n" % name)
        f.write("# Plain ASCII output from Sage\n\n")
        f.write("%s =\n%s\n\n" % (name, P))
```
